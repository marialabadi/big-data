1. المقدمة (Introduction)يهدف هذا العمل إلى تطبيق تقنيات معالجة وتحليل البيانات باستخدام أدوات الذكاء الاصطناعي وتحليل البيانات الحديثة (Power BI). يعتمد المشروع على مجموعة بيانات خاصة بمستشعرات إنذار الحريق والبيئة المحيطة (Cleaned_Dataset) بهدف استخراج مؤشرات الأداء الرئيسية (KPIs) ودعم اتخاذ القرار.   2. معالجة وتطهير البيانات (Data Preprocessing & Cleaning via Power Query)قبل الشروع في عملية التصور البصري، تمت معالجة البيانات بدقة داخل محرر Power Query لضمان خلوها من الأخطاء وضمان جودتها، وشملت الخطوات التالية:   ترقية الرؤوس (En-تêtes promus): ضبط السطر الأول ليصبح أسماءً واضحة للأعمدة الرئيسية مثل (UTC, Temperature[C], Humidity[%], TVOC[ppb], eCO2[ppm], وغيرها).   إزالة التكرار (Doublons supprimés): تنقية الجدول من السجلات المكررة لضمان سلامة التحليلات الإحصائية.   فلترة الأخطاء والفراغات (Lignes filtrées): معالجة السجلات وتصفية القيم الشاذة، حيث سجلت الأعمدة نسبة نجاح بـ 100% وخلو تام من القيم الفارغة أو التالفة (0% أخطاء، 0% فراغ).   تعديل أنواع البيانات (Type modifié): تحويل كل متغير إلى نوع البيانات البرمجي المناسب (أعداد صحيحة للمتغير الزمنية والغازات، وأعداد عشرية لدرجات الحرارة والرطوبة).   هيكلة الجدول (Colonnes supprimées): التخلص من الأعمدة الزائدة وغير الضرورية مثل الفهارس الافتراضية (Unnamed: 0) لتحسين أداء النموذج.   3. التصور البصري ولوحة القيادة (Dashboard & Visualizations)تم تصميم لوحة قيادة تفاعلية متكاملة (Dashboard) على Power BI Desktop لتلخيص مخرجات التحليل، واحتوت على العناصر البصرية التالية:   مخطط الأعمدة الزمنية (Bar Chart): يوضح التوزيع الزمني لقياسات المستشعرات عبر متغير الزمن (UTC).مخطط الحلقات النسبية (Donut Chart): يعرض إجمالي الأحداث (التي بلغت 45 ألف حدث) وتوزيع حالات الإنذار المرتبطة بالحريق (Fire Alarm) والمتغيرات المرتبطة بها.   مخططات التشتت والتحليل المتقدم (Scatter & Analysis Plots): لدراسة العلاقة والارتباط بين الغازات العضوية المتطايرة (TVOC) وثاني أكسيد الكربون (eCO2) مع حالات الطوارئ وإنذار الحريق.   4. الخاتمة (Conclusion)بنجاح هذه المراحل، أصبحت البيانات نظيفة، منظمة، وجاهزة بالكامل للاستخدام في نماذج التعلم الآلي والبيانات الضخمة (Big Data)، مما يبرز كفاءة تطبيق أدوات الذكاء الاصطناعي في إدارة وتحليل البيانات الكبيرة.
